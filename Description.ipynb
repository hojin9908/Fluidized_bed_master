# **KERNEL_DEM_coupling3D_dem**<br>
DEM
$$
Pressure\; Gradient\\
tmpx = \frac{m_a}{\rho_a} \frac{\sum_j{\frac{m_j}{\rho_j}P1\_sph[j].pgf\_x W_{aj}}}{m_a \sum_j{\frac{m_j}{\rho_j}W_{aj}}}\\
tmpy = \frac{m_a}{\rho_a} \frac{\sum_j{\frac{m_j}{\rho_j}P1\_sph[j].pgf\_y W_{aj}}}{m_a \sum_j{\frac{m_j}{\rho_j}W_{aj}}}\\
tmpz = \frac{m_a}{\rho_a} \frac{\sum_j{\frac{m_j}{\rho_j}P1\_sph[j].pgf\_z W_{aj}}}{m_a \sum_j{\frac{m_j}{\rho_j}W_{aj}}}\\[1em]
Drag\; Force\\
uxijf = \frac{\sum_j{u_{x,j}\frac{m_j}{\rho_j}W_{aj}}}{\sum_j{\frac{m_j}{\rho_j}W_{aj}}}-u_{x,a}\\
uyijf = \frac{\sum_j{u_{y,j}\frac{m_j}{\rho_j}W_{aj}}}{\sum_j{\frac{m_j}{\rho_j}W_{aj}}}-u_{y,a}\\
uzijf = \frac{\sum_j{u_{z,j}\frac{m_j}{\rho_j}W_{aj}}}{\sum_j{\frac{m_j}{\rho_j}W_{aj}}}-u_{z,a}\\
mag\_uijf = \sqrt{uxijf^2 + uyijf^2 + uzijf^2}\\
Ref = \frac{D_a(mag\_uijf)\epsilon_a}{\mu_i}\\
Cdf =  (0.63+\frac{4.8}{\sqrt{Ref}})^2\\
betaf = 3.7-0.65*\exp{-(\frac{1.5-log10(Ref)}{2})^2}\\
dfxf = \frac{1}{8}(Cdf) \rho_f \pi D_a^2(uxijf)(mag\_uijf) \epsilon_a^{1-betaf} \frac{1}{m_a}\\
dfyf = \frac{1}{8}(Cdf) \rho_f \pi D_a^2(uyijf)(mag\_uijf) \epsilon_a^{1-betaf} \frac{1}{m_a}\\
dfzf = \frac{1}{8}(Cdf) \rho_f \pi D_a^2(uzijf)(mag\_uijf) \epsilon_a^{1-betaf} \frac{1}{m_a}\\[1em]
Force \; Update\\
P3\_dem.ftotalx = P3.ftotalx + dfxf + tmpx\\
P3\_dem.ftotaly = P3.ftotaly + dfyf + tmpy\\
P3\_dem.ftotalz = P3.ftotalz + dfzf + tmpz\\
SP1\_dem.Fdx\_df = dfxf\\
SP1\_dem.Fdy\_df = dfyf\\
SP1\_dem.Fdz\_df = dfzf
$$
